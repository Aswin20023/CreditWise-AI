# 9. Demographic Analysis

```python
fig = px.histogram(
    df,
    x="AGE",
    nbins=30,
    title="Age Distribution of Credit Card Clients"
)

fig.show()
```

```python
gender = df["SEX"].replace({
    1: "Male",
    2: "Female"
})

fig = px.pie(
    names=gender.value_counts().index,
    values=gender.value_counts().values,
    title="Gender Distribution"
)

fig.show()
```

```python
education = df["EDUCATION"].replace({
    1: "Graduate School",
    2: "University",
    3: "High School",
    4: "Others",
    5: "Unknown",
    6: "Unknown",
    0: "Unknown"
})

fig = px.bar(
    x=education.value_counts().index,
    y=education.value_counts().values,
    title="Education Distribution"
)

fig.show()
```

```python
marriage = df["MARRIAGE"].replace({
    1: "Married",
    2: "Single",
    3: "Others",
    0: "Unknown"
})

fig = px.bar(
    x=marriage.value_counts().index,
    y=marriage.value_counts().values,
    title="Marriage Status"
)

fig.show()
```

# 11. Correlation Analysis

```python
corr = df.corr(numeric_only=True)

fig = px.imshow(
    corr,
    text_auto=".2f",
    aspect="auto",
    title="Correlation Heatmap"
)

fig.show()
```

# 12. Key Findings

## Dataset Summary

- Total records: 30,000
- Total features: 25
- Missing values: None
- Duplicate records: None

## Target Distribution

- No Default: ~77.9%
- Default: ~22.1%

The dataset is moderately imbalanced.

## Correlation

- Repayment history (PAY_0 to PAY_6) shows stronger correlation with default than demographic variables.
- Bill amount variables are strongly correlated with each other.
- Payment amount variables also exhibit positive correlation.

## Initial Observations

- The dataset is clean.
- No missing-value imputation is required.
- Repayment behaviour appears to be the strongest predictor of default.
